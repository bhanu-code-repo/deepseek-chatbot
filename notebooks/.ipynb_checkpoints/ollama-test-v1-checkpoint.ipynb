```python
# Import package
import ollama

# LLM model
llm_model = "deepseek-r1:8b"
```

```python
# Helper function 
def format_markdown(content):
    """Format the given text in markdown with a styled box."""
    # title = f"#{response['title']}"
    # content = f"🚀 {response['content']}\n\n"
    
    # Add note with smaller font
    note = "⚠️  *Note*: This information is for demonstration purposes only."
    content += f"<small>{note}</small>\n\n"
    
    # Add attribution line
    # content += "👨💻 By [Your Name](https://example.com)"
    
    return f"{content}"

def chat_with_agent(user_prompt: str) -> None:
    response = ollama.chat(
        model=llm_model,
        messages=[
            {"role": "user", "content": f"{user_prompt}"},
        ],
    )
    print(response["message"]["content"])
```

```python
chat_with_agent("captial of the UK?")
```
